```python
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()
```

```python
# creating a namespace 

namespace = ('users', 'U1')
```

```python
# put — create a memory
# store.put(namespace, key, value)

namespace = ("users", "U1")

store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})

namespace2 = ("users", "U2")
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid-style navigation"})
```

```python
# get — fetch one specific memory
# store.get(namespace, key)

item = store.get(namespace, "1")
print(item.value) 
```

```python
item = store.get(namespace2, "2")
print(item.value) 
```

```python
# search — fetch all memories in a namespace
# store.search(namespace)

items = store.search(namespace)
for item in items:
    print(item.value)
```

```python
items = store.search(namespace2)
for item in items:
    print(item.value)
```
