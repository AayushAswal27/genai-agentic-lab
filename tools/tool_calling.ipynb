```python
import warnings
warnings.filterwarnings("ignore")
```

```python
@tool
def multiply(a: int, b: int) -> int:
    """Given two numbers a and b, this tool returns their product."""
    return a * b
```

```python
result = multiply.invoke({"a": 3, "b": 4})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
import json
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="conversational",
    provider="auto",
)

model = ChatHuggingFace(llm=llm)
```

```python
llm_with_tools = model.bind_tools([multiply])
```

```python
result = llm_with_tools.invoke("Hi, how are you?")
print(result.content)
print(result.tool_calls)
# []  (empty — no tool needed)
```

```python
result1 = llm_with_tools.invoke("can you multiply 69 with 67")
print(result1.content) # ''  (EMPTY — because it wants to use a tool instead of answering directly)
print(result1.tool_calls)
```

```python
result2 = llm_with_tools.invoke("Can you multiply 3 with 10?")
tool_call = result2.tool_calls[0]

# Option A: pass just the args
multiply.invoke(tool_call["args"])

# Option B: pass the ENTIRE tool_call → returns a ToolMessage
tool_message = multiply.invoke(tool_call)
print(tool_message)
```

```python
from langchain_core.messages import HumanMessage

query = HumanMessage("Can you multiply 3 with 10?")

messages = [query]

# 1. LLM suggests the tool
result = llm_with_tools.invoke(messages)
messages.append(result)   # append the AIMessage (with tool_calls)

# 2. Execute the tool → get a ToolMessage
tool_result = multiply.invoke(result.tool_calls[0])
messages.append(tool_result)   # append the ToolMessage

# 3. Send the FULL history back → LLM gives the final natural answer
final = llm_with_tools.invoke(messages)
print(final.content)
# "The product of 3 and 10 is 30."
```
